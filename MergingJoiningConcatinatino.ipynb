```python
import numpy as np
import pandas as pd
```

```python
data = {
    "emp_id": [101, 102, 103, 104, 105, 106],
    "name": ["Ahnaf", "Rahim", "Karim", "Sadia", "Nusrat", "Tanvir"],
    "department": ["IT", "HR", "CS", "Finance", "IT", "EEE"],   
}
data2 = {
    "emp_id": [101, 102, 103, 104, 107, 108],
    "salary": [50000, 46000, 48000, 61000, 53000, 70000],
    "bonus": [5000, 4000, 3000, 6000, 4500, 8000]
}
df = pd.DataFrame(data)
df2 = pd.DataFrame(data2)
```

```python
df
```

```python
df2
```

```python
pd.merge(df,df2,on="emp_id",how= 'inner' )
```

```python
pd.merge(df,df2,on="emp_id",how= 'outer' )
```

```python
pd.merge(df,df2,on="emp_id",how= 'left')
```

```python
pd.merge(df,df2,on="emp_id",how= 'right')
```

```python
#concatination
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2],axis=1)
```

```python
#joining data frames

# First DataFrame
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df2.join(df1)
```
