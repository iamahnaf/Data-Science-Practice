```python
import openslide
import matplotlib.pyplot as plt

# 1. Load the SVS slide
slide = openslide.open_slide("path_to_your_file.svs")

# 2. View basic properties and dimensions
print(f"Total levels: {slide.level_count}")
print(f"Level 0 (highest resolution) dimensions: {slide.dimensions}")

# 3. Generate and display a low-resolution thumbnail to "see" the whole image
thumbnail = slide.get_thumbnail(size=(600, 600))
plt.imshow(thumbnail)
plt.title("SVS Slide Thumbnail")
plt.axis("off")
plt.show()

# 4. Extract and see a specific high-resolution patch (region)
# Arguments: (x_coord, y_coord), level_index, (width, height)
patch = slide.read_region((5000, 5000), 0, (500, 500))
# Convert RGBA to RGB for display
patch = patch.convert("RGB")
patch.show()

```
