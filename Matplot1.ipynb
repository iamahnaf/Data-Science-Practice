```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y = x**3
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel('X axis')
plt.ylabel('Y axis')
```

```python
plt.subplot(2,2,1)
plt.plot(y**2,x**5)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

```python
#object oriented way
fig = plt.figure(figsize=(20,8),dpi=100)
axis1=fig.add_axes([0.1,0.1,.4,.5])
axis1.plot(x,y)

axis2=fig.add_axes([0.3,0.4,.3,.3])
axis2.plot(x,y)
```
