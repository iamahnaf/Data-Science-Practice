```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
#which team won the most
#value_counts() for counting koi bar nam asche team gular
match_wins = df['match_winner'].value_counts()
match_wins
```

```python
#bar plot
sns.barplot(y=match_wins.index,x=match_wins.values,palette='viridis')
plt.title('Most match win by team')
```

```python
#Toss decision trends
sns.countplot(x = df['toss_decision'])
plt.title('Toss decision Trends')
```

```python
#Toss winner vs match winner
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
print(count)
percentage = (count * 100) / df.shape[0]
percentage.round(2)
```

```python
#player of the match award
countPlayer=df['player_of_the_match'].value_counts().head(10)
countPlayer
```

```python
#two top scorer
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
```

```python
high.plot(kind='barh')
```

```python
#ten bowling figures
df['Highest_wickets'] = df['best_bowling_figure'].apply(lambda x: x.split('--')[0])
df['Highest_wickets'] = df['Highest_wickets'].astype(int)
df.head()
```

```python
topballers = df.groupby('best_bowling')['Highest_wickets'].sum().sort_values(ascending=False).head(10)
```

```python
topballers.plot(kind='barh')
```

```python
#venue analysis
venue_count = df['venue'].value_counts()
venue_count
```

```python
sns.barplot(y=venue_count.index, x=venue_count.values)
```

```python
#Highest margin by runs
Hmargin=df[df['won_by'] == 'Runs'].sort_values(by='margin',ascending=False).head()[['match_winner','margin']]
Hmargin.head(1)
```

```python
#who has the highest score?
df.sort_values(by='highscore',ascending=False).head(1)[['top_scorer','highscore']]
```
